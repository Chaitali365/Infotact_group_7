# GeoPulse — Day 2
## Exploratory Data Analysis (EDA)

### Objective

Understand the structure, distribution, temporal patterns, and geographic
characteristics of the synthetic mobility and retail datasets.

### Analysis Areas

- Mobility activity distribution
- Movement type analysis
- Device type analysis
- Hourly mobility patterns
- Daily mobility patterns
- Store location analysis
- Geographic coverage

```python
import pandas as pd
import numpy as np

mobility_df = pd.read_csv("mobility_pings.csv")
stores_df = pd.read_csv("stores.csv")
zones_df = pd.read_csv("zones.csv")

mobility_df["timestamp"] = pd.to_datetime(mobility_df["timestamp"])

print("Datasets loaded successfully.")

print("Mobility Dataset:", mobility_df.shape)
print("Stores Dataset:", stores_df.shape)
print("Zones Dataset:", zones_df.shape)

print("\nMobility Columns:")
print(mobility_df.columns.tolist())

print("\nStores Columns:")
print(stores_df.columns.tolist())

print("\nZones Columns:")
print(zones_df.columns.tolist())

#Movement Type Analysis
movement_counts = mobility_df["movement_type"].value_counts()

print("Movement Type Distribution:")
print(movement_counts)

#Percentage Distribution
movement_percentage = (
    mobility_df["movement_type"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)

print("Movement Type Percentage:")
print(movement_percentage)

#Device Type Count
device_counts = mobility_df["device_type"].value_counts()

print("Device Type Distribution:")
print(device_counts)

#Device Type Percentage
device_percentage = (
    mobility_df["device_type"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)

print("Device Type Percentage:")
print(device_percentage)

#Hour-wise Mobility Pattern
hourly_mobility = (
    mobility_df["hour"]
    .value_counts()
    .sort_index()
)

print("Hourly Mobility Distribution:")
print(hourly_mobility)

#peak hour
peak_hour = hourly_mobility.idxmax()
peak_hour_count = hourly_mobility.max()

print(f"Peak mobility hour: {peak_hour}:00")
print(f"Mobility pings during peak hour: {peak_hour_count}")

#. Day-of-Week Analysis
daily_mobility = mobility_df["day_of_week"].value_counts()

print("Day-of-Week Mobility Distribution:")
print(daily_mobility)

#chronological view
day_order = [
    "Monday",
    "Tuesday",
    "Wednesday",
    "Thursday",
    "Friday",
    "Saturday",
    "Sunday"
]

daily_mobility_ordered = (
    mobility_df["day_of_week"]
    .value_counts()
    .reindex(day_order)
)

print(daily_mobility_ordered)

#Movement Type × Hour
movement_by_hour = pd.crosstab(
    mobility_df["hour"],
    mobility_df["movement_type"]
)

print(movement_by_hour)

#percentage version
movement_by_hour_percentage = pd.crosstab(
    mobility_df["hour"],
    mobility_df["movement_type"],
    normalize="index"
).mul(100).round(2)

print(movement_by_hour_percentage)

#Store Dataset Analysis
print("Store Types:")
print(stores_df["store_type"].value_counts())

print("\nAverage Daily Capacity:")
print(round(stores_df["daily_capacity"].mean(), 2))

print("\nTotal Daily Capacity:")
print(stores_df["daily_capacity"].sum())

#Geographic Coverage
print("Mobility Latitude Range:")
print(mobility_df["latitude"].min(), "to", mobility_df["latitude"].max())

print("\nMobility Longitude Range:")
print(mobility_df["longitude"].min(), "to", mobility_df["longitude"].max())

print("\nStore Latitude Range:")
print(stores_df["latitude"].min(), "to", stores_df["latitude"].max())

print("\nStore Longitude Range:")
print(stores_df["longitude"].min(), "to", stores_df["longitude"].max())

#Business-Oriented Summary
print("========== GEOPULSE DAY 2 EDA SUMMARY ==========")

print(f"Total mobility pings: {len(mobility_df)}")
print(f"Total stores: {len(stores_df)}")
print(f"Total zones: {len(zones_df)}")

print(f"\nMost common movement type: {mobility_df['movement_type'].mode()[0]}")
print(f"Most common device type: {mobility_df['device_type'].mode()[0]}")

peak_hour = hourly_mobility.idxmax()
peak_hour_count = hourly_mobility.max()

print(f"Peak mobility hour: {peak_hour:02d}:00 - {peak_hour:02d}:59")
print(f"Peak hour pings: {hourly_mobility.max()}")

print(f"\nHighest mobility day: {daily_mobility_ordered.idxmax()}")
print(f"Highest day pings: {daily_mobility_ordered.max()}")

print(f"\nTotal store daily capacity: {stores_df['daily_capacity'].sum()}")
print("=================================================")
```
