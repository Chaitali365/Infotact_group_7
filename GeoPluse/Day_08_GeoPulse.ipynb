# GeoPulse — Day 8
## Snowflake Data Warehouse Integration

## Objective

Integrate the GeoPulse analytical datasets into a Snowflake data warehouse and validate the loaded data using SQL.

## Work Completed

- Created the GeoPulse Snowflake database
- Created RAW and ANALYTICS schemas
- Loaded GeoPulse datasets into Snowflake
- Validated table availability
- Validated row counts
- Performed data quality checks
- Performed mobility aggregation using Snowflake SQL
- Analyzed store opportunity metrics
- Analyzed store performance metrics
- Created an analytical view for downstream reporting
