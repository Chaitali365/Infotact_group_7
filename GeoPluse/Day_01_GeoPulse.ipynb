```python
# ============================================
# GEOPULSE - SYNTHETIC MOBILITY DATA
# ============================================

import pandas as pd
import numpy as np
from datetime import datetime, timedelta

# Reproducible random data
np.random.seed(42)

# Number of GPS pings
n = 5000

# Generate IDs
ping_ids = [f"P{str(i).zfill(5)}" for i in range(1, n + 1)]
user_ids = [f"U{str(np.random.randint(1, 1001)).zfill(4)}" for _ in range(n)]

# Fictional city center
center_lat = 28.6200
center_lon = 77.2100

# Generate GPS coordinates around the city
latitudes = np.random.normal(center_lat, 0.025, n)
longitudes = np.random.normal(center_lon, 0.025, n)

# Generate timestamps over 30 days
start_date = datetime(2026, 1, 1)

timestamps = [
    start_date + timedelta(
        days=int(np.random.randint(0, 30)),
        hours=int(np.random.randint(0, 24)),
        minutes=int(np.random.randint(0, 60)),
        seconds=int(np.random.randint(0, 60))
    )
    for _ in range(n)
]

# Sort timestamps
timestamps = sorted(timestamps)

# Movement types
movement_types = np.random.choice(
    ["Walking", "Driving", "Public Transport"],
    size=n,
    p=[0.50, 0.30, 0.20]
)

# Device types
device_types = np.random.choice(
    ["Mobile", "Tablet"],
    size=n,
    p=[0.90, 0.10]
)

# Create DataFrame
mobility_df = pd.DataFrame({
    "ping_id": ping_ids,
    "anonymous_user_id": user_ids,
    "timestamp": timestamps,
    "latitude": latitudes.round(6),
    "longitude": longitudes.round(6),
    "device_type": device_types,
    "movement_type": movement_types
})

# Add date and time-related columns
mobility_df["date"] = mobility_df["timestamp"].dt.date
mobility_df["hour"] = mobility_df["timestamp"].dt.hour
mobility_df["day_of_week"] = mobility_df["timestamp"].dt.day_name()

# Save CSV
mobility_df.to_csv("mobility_pings.csv", index=False)

print("mobility_pings.csv created successfully!")
print("Rows:", len(mobility_df))
print("\nFirst 5 rows:")
display(mobility_df.head())
```
